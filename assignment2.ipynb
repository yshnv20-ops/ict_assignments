 Create a bill calculator where the user enters the number of units consumed. ₹5 per unit for the first 100 units ₹7 per unit for the next 100 units (from 101 to 200) ₹10 per unit for any units above 200 Calculate and display the total electricity bill. Sample Input: Enter units consumed: 245 Sample Output: Total Bill Amount: ₹1650  


```python
units = int(input("Enter the number of units consumed: "))
bill = 0
if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
else:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10
print(f"Total Electricity Bill = ₹{bill}")
```
